## Exercise 1 : Number representation in computer

> Let us consider a floating-point representation of normalised numbers in binary ([1,2])
>
> $$
\begin{aligned}
x &= (-1)^\sigma \times (1.b_1 b_2 \dots b_p)_2 \times 2^{E-B(E_*)}, \qquad{}  (1)\\
E &= (e_1 e_2 \dots e_{E_*})_2, \qquad{} B(E_*) = 2^{E_*-1}-1,
\end{aligned}
$$
>
> where $p$, the size of the mantissa, and $E_*$, that of the exponent, are given positive integers and the digits $\sigma$,  $b_i$ and $e_i$ are $0$ or $1$ (we work in binary).
>
>The function $fl(x)$ denotes the rounding of a number $x$ obtained in representation (1) by truncating the mantissa to $p$ significant digits.

> In everything that follows, whenever we consider an arbitrary real number $x$, we shall assume that $\vert x\vert\leq M_{p,E_*}$.

> For a representation of the form (1) (with $p$ and $E_*$ fixed), one commonly calls machine epsilon *the smallest real number which, added to 1, gives a floating-point number strictly greater than one*. In a mathematically more precise way, we define
>
>$$\epsilon_M := \inf_{fl(1+x)>1} x.$$

```python
def f(x,y):
    """ Definition of the function f """
    return 9*x**4-y**4+2*y**2
```

**Single precision**

```python
# mp.prec is the size of the mantissa (in bits) used by floating-point numbers,
# including the leading bit.

# A mantissa of 24 (23 + 1) bits corresponds to the ieee 32 standard ("single precision")
mp.prec = 24

# mp.mpf makes it possible to define floating-point numbers whose mantissa has mp.prec bits
# (potentially different from the default one)
x = mp.mpf('40545')
y = mp.mpf('70226')

print(f"Evaluation of the function for the floating-point type with a mantissa of {mp.prec} bits")
print(f"f(x,y) = ", f(x,y))
```

**Double precision**

This is the default precision of floating-point numbers in Python.

```python
# This formulation (with a . at the end) is the default one for real numbers,
# that is to say floating-point numbers in the ieee 64 standard ("double precision")
x = 40545.
y = 70226.

print(f"Evaluation of the function for the type {type(x).__name__}")
print(f"f(x,y) = ", f(x,y))
```

```python
# A mantissa of 53 (52 + 1) bits corresponds to the ieee 64 standard ("double precision")
mp.prec = 53
x = mp.mpf('40545')
y = mp.mpf('70226')

print(f"Evaluation of the function for the floating-point type with a mantissa of {mp.prec} bits")
print(f"f(x,y) = ", f(x,y))
```

**Integer**

**Quadruple precision**

```python
# A mantissa of 113 (112 + 1) bits corresponds to the ieee 128 standard ("quadruple precision")
mp.prec = 113
x = mp.mpf('40545')
y = mp.mpf('70226')

print(f"Evaluation of the function for the floating-point type with a mantissa of {mp.prec} bits")
print(f"f(x,y) = ", f(x,y))
```

**Arbitrary precision**

```python
# You can change the value of mp.prec below (and hence the size of the mantissa)
mp.prec = 65
x = mp.mpf('40545')
y = mp.mpf('70226')

print(f"Evaluation of the function for the floating-point type with a mantissa of {mp.prec} bits")
print(f"f(x,y) = {f(x,y)}")
```

## Exercise 3 : Stability and conditioning

For a map $\mathcal{P}:\mathbb{R}^n\to \mathbb{R}$, we recall the following definitions. In what follows $\epsilon>0$ is arbitrarily small.

- The condition number of the problem of evaluating $\mathcal{P}(x)$ is the smallest scalar $\kappa$ of amplification of the relative error on the result

    $$ \frac{|\hat{x}_i-x_i|}{|x_i|} \le \epsilon \quad\text{for all $i$} \quad \Rightarrow \quad \frac{|\mathcal{P}(\hat{x})-\mathcal{P}(x)|}{|\mathcal{P}(x)|} \le \kappa \epsilon. $$

    A problem is well conditioned if $\kappa$ is small.

- Considering an algorithm $\hat{\mathcal{P}}(x)$ which approximates $\mathcal{P}(x)$, its backward stability constant is the smallest scalar $C$ of amplification of the error on the data, i.e.

    $$ \exists \hat{x}\quad \text{satisfying} \quad \mathcal{P}(\hat{x}) = \hat{\mathcal{P}}(x) \quad \text{such that}\quad \frac{|\hat{x}_i-x_i|}{|x_i|} \le  C \epsilon_M \quad \text{for all $i$}. $$

    An algorithm is stable if $C$ is small.
