## Exercise 1: Floating-point representation

Consider the following binary representation of non-zero normalised floating-point numbers:

$$
\begin{aligned}
x &= (-1)^\sigma (1.b_1b_2\ldots b_p)_2\,2^{E-B(E_*)}, \qquad\qquad (1)\\
E &= (e_1e_2\ldots e_{E_*})_2,
\qquad B(E_*)=2^{E_*-1}-1,
\end{aligned}
$$

where $p$ is the number of fraction bits, $E_*$ is the number of exponent bits, and
$\sigma,b_i,e_i\in\{0,1\}$. Thus the significand has $p+1$ bits when the implicit
leading bit is included.

The function $fl(x)$ denotes the number obtained by truncating the mantissa of $x$ after
$p$ fraction bits. We exclude overflow and underflow.

In what follows, all real numbers are assumed to lie in the normal range, so that no
overflow or underflow occurs.

For fixed $p$ and $E_*$, the **machine epsilon** is commonly described as the smallest
real number which, when added to $1$, produces a floating-point number strictly greater
than $1$. More precisely, we define

$$
\epsilon_M:=\inf_{fl(1+\eta)>1}\eta.
$$

## Exercise 3: Stability and conditioning

Let $\mathcal P:\mathbb R^d\to\mathbb R$ and assume $\mathcal P(x)\ne0$. Its local
componentwise relative condition number is

$$
\kappa(x)=
\lim_{\eta\downarrow0}
\sup_{|h_i|\le\eta|x_i|}
\frac{|\mathcal P(x+h)-\mathcal P(x)|}
     {\eta|\mathcal P(x)|}.
$$

An algorithm $\widehat{\mathcal P}$ is backward stable at $x$ if there exists a
perturbed input $\widehat x$ such that

$$
\widehat{\mathcal P}(x)=\mathcal P(\widehat x),
\qquad
|\widehat x_i-x_i|\le C\epsilon_M |x_i|,
$$

with a moderate constant $C$. Fixed exact constants appearing in the problem are not
included among the perturbable inputs.

### Question 1

$$

$$

$$
