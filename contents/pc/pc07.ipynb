## Exercise 2: Consistency + stability $\Rightarrow$ convergence


>Consider $f\in\mathcal{C}^0(\mathbb{R}^2,\mathbb{R})$, locally Lipschitz with respect to its second variable, and the Cauchy problem
>
>\begin{equation*}
\left\{
\begin{aligned}
&y'(t)=f(t,y(t)),\\
&y(t_0)=y_0,
\end{aligned}
\right.
\qquad{} (1)
\end{equation*}
>
>which we want to solve numerically. In this exercise, we only consider the explicit Euler method,
>
>\begin{equation*}
y_{n+1}=y_n+(t_{n+1}-t_n)f(t_n,y_n).
\end{equation*}
>
>The notions and results below also extend to other methods studied in the course. In particular, we will prove part of the *Lax theorem*: for a consistent scheme, stability is a necessary and sufficient condition for convergence.
>
>Given a subdivision $t_0<t_1<\cdots<t_N$, we write $\Delta t_n=t_{n+1}-t_n$.
>
>*To simplify some questions, you may assume that $f$ is of class $\mathcal{C}^1$.*


### Question 1: Consistency

>**Definitions:** Let $y$ be a solution of (1) defined on $[t_0,t_0+T]$, and let $t_0<t_1<\cdots<t_N=t_0+T$ be a subdivision. For $0\leq n<N$, define
>
>\begin{equation*}
\varepsilon_n=y(t_{n+1})-y(t_n)-\Delta t_n f(t_n,y(t_n)),
\end{equation*}
>
>that is, the error obtained when the numerical scheme is applied to the exact solution $y$ at the points $t_n$. The **consistency error** is
>
>\begin{equation*}
E=\sum_{n=0}^{N-1}|\varepsilon_n|.
\end{equation*}
>
>The explicit Euler method is called **consistent** if
>
>\begin{equation*}
E\underset{\Delta t\to0}{\longrightarrow}0,
\qquad
\Delta t=\max_{0\leq n<N}\Delta t_n.
\end{equation*}

1) Assume that $f$ is of class $\mathcal{C}^1$. Show that the consistency error is $O(\Delta t)$, and therefore that the explicit Euler method is consistent.

2) Show that the explicit Euler method is still consistent if $f$ only satisfies the Cauchy assumptions: $f$ is continuous and locally Lipschitz with respect to $y$.


### Question 2: Stability

>**Definition:** The explicit Euler method is called **stable** if there exists a constant $C>0$, bounded as $\Delta t\to0$, such that for every sequence $(z_i)_{0\leq i\leq N}$ satisfying
>
>\begin{equation*}
z_{n+1}=z_n+\Delta t_n f(t_n,z_n)+\delta_n,
\qquad 0\leq n<N,
\end{equation*}
>
>we have
>
>\begin{equation*}
\max_{0\leq n\leq N}|y_n-z_n|
\leq
C\left(
|y_0-z_0|
+
\sum_{n=0}^{N-1}|\delta_n|
\right).
\end{equation*}

## Exercise 3: Implicit Euler and different notions of stability


Below, we give an implementation of the implicit Euler method for scalar equations of the form
$$
\left\{
\begin{array}{rl}
y'(t)&=f(t,y(t)),\\
y(0)&=y_0.
\end{array}
\right.
$$


```python
def Newton_scalaire(f, df, x0, tol=1.e-12, it_max=100):
    """
    Compute an approximate solution of f(x)=0 with Newton's method.
    ----------   
    Parameters:
    f      : function whose zero is sought
    df     : derivative of f
    x_0    : initial value
    tol    : stopping tolerance based on |f(x_n)| < tol 
    it_max : maximum number of iterations
    
    Returns:
    last iterate x_n
    """ 

    n = 0
    x = x0
    fx = f(x)
    err = np.abs(fx)
    
    while  err > tol  and  n < it_max :
        x = x - fx/df(x)
        n += 1
        fx = f(x)
        err = np.abs(fx)
    
    if err > tol:
        print("Warning: the prescribed tolerance was not reached.")
        
    return x
```

```python
def Euler_implicite(f, df, y0, T_f, N, tol=1.e-12, it_max=100):
    """
    Compute the successive iterates y_n obtained with the implicit Euler method and a constant time-step.
    ----------   
    Parameters:
    f      : right-hand side, depending on y and t
    df     : derivative of f with respect to y, depending on y and t
    y_0    : value at time t_0
    T_f    : final time
    N      : number of time-steps (dt = T_f/N)
    tol    : residual tolerance for Newton's method 
    it_max : maximum number of Newton iterations
    
    Returns:
    tab_y  : array containing the values y_n at each iteration
    """ 
    
    dt = T_f / N
    tab_y = np.zeros(N+1)
    tn = 0
    tab_y[0] = y0
    for n in np.arange(0,N):
        def g(y):
            return y - tab_y[n] - dt * f(tn,y)    
        def dg(y):
            return 1 - dt * df(tn,y)
        tab_y[n+1] = Newton_scalaire(g, dg, tab_y[n], tol, it_max)
        tn += dt
        
    return tab_y
```
