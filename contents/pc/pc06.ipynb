### Question 2

Test the bisection algorithm given below on
$$
g(x)=x^3-2,
$$
with $a=1$ and $b=2$. Plot the error $\epsilon_n=|x_n-x_*|$ using a suitable scale, together with the error bound obtained in the previous question.


```python
def bissection(f, a, b, tol=1.e-12, N_max=500):
    """
    Computes the successive iterates x_n obtained with the bisection method on [a,b].
    ----------   
    Parameters:
    f     : function whose zero is sought
    a,b   : endpoints of the interval in which a zero is sought
    tol   : stopping tolerance based on |a-b| < tol
    N_max : maximum number of iterations
    
    Returns:
    x : array containing the values x_n = (a_n+b_n)/2 at each iteration
    """  
                
    if f(a)*f(b) > 0:
        print("f(a) and f(b) have the same sign!")
        return
        
    x = np.zeros(N_max+1)    
    n = 0                
    x[0] = (a+b)/2        
    fa = f(a)
    fx = f(x[0])
    while  np.abs(fx) > tol  and  n < N_max :  
        if fa * fx <= 0 :
            b = x[n] 
        else:
            a = x[n]
            fa = fx
        n += 1
        x[n] = (a+b)/2
        fx = f(x[n])
            
    if np.abs(fx) > tol:
        print("Warning: the prescribed tolerance was not reached")
        
    return x[:n+1] 
```
