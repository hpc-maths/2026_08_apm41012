## Exercise 1: From polynomial interpolation to quadrature convergence

>Let $a<b$ and let $x_0,\ldots,x_n$ be distinct nodes in $[a,b]$. As in PC 2, write
>
>$$
\ell_j(x)=\prod_{\substack{r=0\\r\ne j}}^n\frac{x-x_r}{x_j-x_r},
\qquad
p_n(f)(x)=\sum_{j=0}^n f(x_j)\ell_j(x),
\qquad
\Pi_{n+1}(x)=\prod_{j=0}^n(x-x_j).
$$
>
>The associated interpolatory quadrature rule is
>
>$$
I_n^{[a,b]}(f)=\int_a^b p_n(f)(x)\,dx.
$$
>
>We use $\|\cdot\|_\infty$ for the uniform norm on $[a,b]$.

### Question 1

Recall and compare the error estimates from the lectures for the following quadrature rules:

- Gauss–Legendre, using $n+1$ quadrature points;
- Clenshaw–Curtis, using $n+1$ quadrature points;
- the trapezoidal rule, using $n+1$ quadrature points, where $n$ is also the number of subintervals;
- composite Gauss–Legendre, using two quadrature points per subinterval and $(n+1)/2$ subintervals, for a total of $n+1$ quadrature points.

*Your answer here.*

>Implementations of the Gauss–Legendre and Clenshaw–Curtis rules with $n+1$ points, and of composite Gauss–Legendre quadrature, are provided below.

```python
def GL(f, a, b, n):
    """
    Approximate the integral of f over [a, b]
    using Gauss-Legendre quadrature with n + 1 points.
    ----------
    Parameters:
    f    : function to integrate
    a, b : integration limits
    n    : number of quadrature points minus one

    Returns:
    Approximation to the integral.
    """
    x, w = p_roots(n+1)
    return np.sum(w * f( x*(b-a)/2+(b+a)/2 ) ) * (b-a)/2
```

```python
def CC(f, a, b, n):
    """
    Approximate the integral of f over [a, b]
    using the supplied CC rule with n + 1 points.
    ----------
    Parameters:
    f    : function to integrate
    a, b : integration limits
    n    : number of quadrature points minus one

    Returns:
    Approximation to the integral.
    """
    theta = (2*np.linspace(0,n,n+1)+1)/(2*n+2)*np.pi
    x     = np.cos(theta)
               
    w = np.ones(n+1)
    for k in range(1,n+1):
        if(k%2==0):
            w += 2.*np.cos(k*theta)/(1.-k**2)
    w *= 2./(n+1)   
    return np.sum(w * f( x*(b-a)/2+(b+a)/2 ) ) * (b-a)/2
```

```python
def GLcomp(f, a, b, n, m):
    """
    Approximate the integral of f over [a, b] using composite
    Gauss-Legendre quadrature with n subintervals and m points per subinterval.
    ----------
    Parameters:
    f    : function to integrate
    a, b : integration limits
    n    : number of subintervals
    m    : number of quadrature points per subinterval

    Returns:
    Approximation to the integral.
    """
    x, w = p_roots(m)
    sub = np.linspace(a, b, n+1)
    I = 0
    for k in range(0,n):
        ak = sub[k]
        bk = sub[k+1]
        I += np.sum(w * f( x*(bk-ak)/2+(bk+ak)/2 ) ) * (bk-ak)/2
    return I
```

```python
# Test
a, b = -1, 1
n = 9
I_GL = GL(np.exp, a, b, n)
I_CC = CC(np.exp, a, b, n)
I_GLcomp = GLcomp(np.exp, a, b, (n + 1) // 2, 2)

print(f"Gauss-Legendre quadrature of exp on [{a}, {b}] with {n + 1} points: {I_GL}")
print(f"Clenshaw-Curtis quadrature of exp on [{a}, {b}] with {n + 1} points: {I_CC}")
print(f"Composite Gauss-Legendre quadrature of exp on [{a}, {b}] with {n + 1} points: {I_GLcomp}")
print("e - 1/e =", np.exp(1) - np.exp(-1))
```
