# The numpy module

The ```numpy``` module makes it possible to manipulate multidimensional arrays efficiently.

Unlike Python lists, a ```numpy``` array can only store elements of the same type.

The data of a ```numpy``` array is stored contiguously in memory, which allows the functions of the module to be efficient when manipulating the data of the array.

## Importing the module

To use the module, simply import it with the ```import``` instruction. It is customary to rename it ```np``` using the ```as``` keyword:

````python
import numpy as np
````

## Main attributes

### Data type

Creating a `numpy` array from a list:

````python
a = np.array([1, 2, 3.5, 4, 5])
print(a)
````

The type of the object referenced by the variable `a` is:

````python
type(a)
````

This type must not be confused with the type of the stored data, which can be obtained through the `dtype` attribute:

````python
print(a.dtype)
````

The type chosen by default depends on the input data; in the previous case it is the 64-bit floating-point type, so as to avoid any loss of precision.

The data type can be specified explicitly (with a possible loss of precision):

````python
b = np.array([1, 2, 3.5, 4, 5], dtype=np.int32)
print(b)
````

The complete list of available types can be found [here](https://numpy.org/doc/stable/user/basics.types.html).

### Number of dimensions, shape and size of a numpy array

The `ndim` attribute gives the number of dimensions:

````python
# one-dimensional array
a = np.array([1, 2, 3, 4, 5, 6])
print(a)
print(a.ndim)
````

````python
# two-dimensional array
b = np.array([[1, 2, 3, 4], [5, 6, 7, 8]])
print(b)
print(b.ndim)
````

The `shape` attribute gives the dimensions of the array:

````python
print(a.shape)
print(b.shape)
````

The `size` attribute gives the size of the array:

````python
print(a.size)
print(b.size)
````

## Creating numpy arrays

The `numpy` module provides a large number of methods to build `numpy` arrays. Here are a few examples of how to use these methods.

The `np.ones` method returns an array containing only 1s:

````python
# Building a 1-dimensional array of size 5 containing only 1s
a = np.ones(5)
print(a)
````

````python
# Building a 2-dimensional array of size 2x3 containing only 1s
b = np.ones((2,3))
print(b)
````

The `np.zeros` method returns an array containing only 0s:

````python
# Building a 1-dimensional array of size 5 containing only 0s
a = np.zeros(5)
print(a)
````

````python
# Building a 2-dimensional array of size 2x3 containing only 0s
b = np.zeros((2,3))
print(b)
````

The `np.arange` method returns an array containing evenly spaced values within a given interval:

````python
# Building a 1-dimensional array containing evenly spaced values
# between 1 and 9 (excluded) with a step of 2
a = np.arange(1,9,2)
print(a)
````

````python
# By default, the start of the interval is 0 and the step is 1
b = np.arange(10)
print(b)
````

The `np.linspace` method returns an array containing `n` evenly spaced values within a given interval:

````python
# Building a 1-dimensional array containing 20 evenly spaced values between 1 and 10
a = np.linspace(1,10, 20)
print(a.size)
````

A complete list of the `numpy` array creation methods can be found [here](https://numpy.org/doc/stable/reference/routines.array-creation.html).

## Indexing and slicing

The `[]` operator gives access to an element of an array. Indexing starts at 0.

````python
## case of a one-dimensional array
a = np.ones(5)
a[2] = 4
print(a)
````

````python
# case of a two-dimensional array
b = np.ones((2,3))
b[1,1] = 4
print(b)
````

````python
# in this case an element can also be accessed this way
b[0][2] = 6
print(b)
````

Negative indices can be used to index from the end of the array:

````python
a = np.ones(5)
a[-1] = 2
print(a)
````

Slicing makes it possible to reference a part of an array:

````python
# referencing a contiguous part
a = np.arange(10)
print("a       = ", a)
print("a[1:4]  = ", a[1:4])
print("a[1:]   = ", a[1:])
print("a[:3]   = ", a[:3])
print("a[2:-2] = ", a[2:-2])
print("a[:]    = ", a[:] )
````

````python
# referencing a non-contiguous part
a = np.arange(10)
print("a         = ", a)
print("a[1:6:2]  = ", a[1:6:2])
print("a[8:2:-2] = ", a[8:2:-2])
````

````python
# examples with multidimensional arrays

# creating a (5x5) array of random integers between 1 and 10
a = np.random.randint(1, 10, (5,5))
print("a = ")
print(a)
# extracting a row
print("\na[1,:] =")
print(a[1,:])

# extracting a column
print("\na[:,1] =")
print(a[:,1])

# extracting a block
print("\na[1:2,3:4] =")
print(a[0:3,2:4])
````

The `numpy` documentation on indexing and slicing is available [here](https://numpy.org/doc/stable/reference/arrays.indexing.html).

## Mutability of numpy arrays

A `numpy` array is a mutable object, which makes it possible to modify one of its elements without creating a new object. The `[]` operator gives access to an element of an array.

````python
a = np.array([1, 2, 3.5, 4, 5])
a[2] = 3
print(a)
````

As a consequence, the assignment operation does not create a copy.

````python
b = a
print("Identifier of a = ", id(a))
print("Identifier of b = ", id(b))
````

If an element of the array referenced by `a` is modified, an element of the array referenced by `b` is also modified, since both variables reference the same `ndarray` object:

````python
print("b = ", b)
a[0] = 6
print("b = ", b)
````

It is possible to create a copy explicitly:

````python
a = np.array([1, 2, 3.5, 4, 5])
b = np.copy(a)
print("Identifier of a = ", id(a))
print("Identifier of b = ", id(b))
````

By combining the assignment operation and slicing, views on a part of an array can be created:

````python
a = np.arange(10)
v = a[1:4]
v[:] = 2
print(a)
````

## Vectorisation

Vectorisation is a powerful capability of `numpy` that makes it possible to express operations directly on whole arrays rather than on their elements. This practice consists in replacing explicit loops by expressions on arrays, which are far more efficient.

This is the reason why `numpy` arrays are more efficient than `python` lists, since the latter require loops to manipulate their elements:

````python
# initialising an array containing the squares of the first n integers
n = 10000
````

````python
%%timeit
# with a python list
l = []
for i in range(n):
    l.append(i**2)
````

````python
# with a list comprehension
%timeit l = [i**2 for i in range(n)]
````

````python
# with a numpy array, the ** operator can be used directly on the array
%timeit a = np.arange(n)**2
````

The `numpy` version is far more efficient, and it is precisely because loops have been avoided that the `numpy` version is faster:

````python
%%timeit 
a = np.empty(n)
for i in range(n):
    a[i] = i**2
````

As far as possible, avoid using loops to manipulate arrays.
