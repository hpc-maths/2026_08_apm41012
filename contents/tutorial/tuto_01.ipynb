# Variables in Python

## The notions of object and variable

### The notion of object

In Python, everything is an object.

An object is an entity that holds data and provides mechanisms called methods to manipulate it.

#### Example

Creating an object of integer type:

````python
10
````

The object ```10``` provides a method named ```bit_length``` which returns the number of bits needed to represent it in binary. It is accessed through the ```.``` character:

````python
(10).bit_length()
````

Note: the number of bits used to encode an integer is dynamic:

````python
(1000).bit_length()
````

### The notion of variable

A variable is a reference to an object.

````python
a = 20
````

This instruction creates the integer object ```20```, then creates the variable ```a``` and binds the variable ```a``` to the object ```20```, which allows the variable ```a``` to manipulate the object ```20```.

````python
a.bit_length()
````

The variable ``a`` can reference another object of the same type:

````python
a = 51
````

This instruction creates the integer object `51` and binds the variable a to the object ```51```.

It can also reference an object of a different type (a character string, for instance):

````python
a = "string"
````

This is why Python is said to use dynamic typing (the type is not bound to the variable but to the object referenced by the variable).

### Main available types

The available basic types are:

* the numeric types:
  * integers (int)
  * floating-point numbers (float)
  * complex numbers (complex)
* booleans (bool)
* character strings (str)

Python also provides many other types such as lists (list), which are sequences of heterogeneous objects; the exhaustive list can be found [here](https://docs.python.org/3.8/library/stdtypes.html#sequence-types-list-tuple-range).

The ```type``` function returns the type of an object or the type of the value of a variable; simply use the type() function:

````python
# integer
type(50)
````

````python
# float
f = 10.
type(f)
````

````python
# complex
c = 1 + 2j
type(c)
````

````python
# boolean
type(False)
````

````python
# character string
ch = "string"
type(ch)
````

````python
# list
l = [2, f, True, ch]
type(l)
````

### Defining a function

A function is a block of instructions that can be called by a name. The syntax for defining a function is the following:

```python
def function_name(list of parameters):
      block of instructions
```

Example:

````python
def show(param):
    print(param)
````

Example calls:

````python
show(6)
````

````python
ch = "a string"
show(ch)
````

A function is an object of type `function`:

````python
type(show)
````

It is possible to bind a variable to a function object:

````python
a = show
a(6)
````

## Variable scope

### Global variable

A variable defined in a cell outside a function or a class is considered global; it is accessible throughout the whole notebook:

````python
a = 12
````

````python
print("a is", a)
````

````python
def fct_01():
    print("In function fct_01, a is", a)
````

````python
fct_01()
````

### Local variable

A variable defined inside a function or a class is local; it is only accessible from the function or the class where it was defined:

````python
def fct_02():
    b = 12
    print("In function fct_02, b is", b)
````

````python
fct_02()
````

Outside the function ```fct_02```, the variable ```b``` is not defined. The following instruction raises an error:

````python
# the variable b is not defined outside the function, the following instruction raises an error
print(b)
````

Be careful: when a variable is assigned in a given context, that variable becomes local to this context and shadows any variable of the same name from the calling context.

````python
def fct_03():
    a = 100
    print("In function fct_03, a is", a)
````

In the function ```fct_03```, even though the local variable ```a``` has the same name as a global variable, the reference to the variable ```a``` designates the local variable:

````python
fct_03()
````

Outside the function, a reference to the variable ```a``` designates the global variable:

````python
print("a = ", a)
````

It is possible to modify a global variable inside a function by using the ```global``` keyword:

````python
def fct_04():
    global a
    a += 100
    print("In function fct_04, a is", a)
````

````python
fct_04()
````

We can check that the global variable ```a``` has indeed been modified:

````python
print("a = ", a)
````

Using this method is not recommended.

## Function arguments

In Python, arguments are passed by reference.

````python
def fct_05(a,b):
    print(a)
    print(b)
````

So it is indeed the objects referenced by the variables given as arguments that are retrieved inside the function:

````python
c = 10
d = 12
fct_05(c,d)
````

Be careful: if a new object is assigned to the variable, the variable becomes local to the function:

````python
def fct_06(a,b):
    a = a + b
    print("Inside function fct_06, a = ", a)
````

````python
a = 10
b = 12
fct_06(a,b)
print("Outside function fct_06, a = ", a)
````

The variable ``a`` that was modified inside the function ``fct_06`` is local to the function.
Outside the function, it does not reference the same object.

Be careful, this behaviour is due to the non-mutable nature of integer objects. It would be different for objects of type list (see the next section on mutability).

To retrieve an object that is local to the function, the function must return a value using the ```return``` keyword:

````python
def fct_07(a,b):
    s = a + b
    return s
````

````python
a = 8
b = 10
print(f"Sum of {a} and {b}: {fct_07(a,b)}")
````

## Mutability

An immutable (or non-mutable) object is an object that cannot be modified after its creation. In Python, this is the case, in particular, for the numeric types, the booleans and the character strings.

Thus, for non-mutable types, using operators such as ```+=```, ```-=```, ... creates a new object. This can be checked using the ```id``` function, which returns a unique identifier of the object given as a parameter (representative of its address in memory).

````python
a = 40
print(f"Identifier of the variable a = {a}: {id(a)}")
a += 5
print(f"Identifier of the variable a = {a}: {id(a)}")
````

Lists are mutable types, which makes it possible to modify or add elements without creating a new list:

````python
l = [2, True, 2.5, 'ch']
print(f"Identifier of the list l = {l}: {id(l)}")
l[2] = 5
print(f"Identifier of the list l = {l}: {id(l)}")
l += [8, 6.8]
print(f"Identifier of the list l = {l}: {id(l)}")
````

As a consequence, inside a function, it is possible to modify a list passed as a parameter:

````python
def fct_08(a):
    a += [1]
````

````python
fct_08(l)
print(l)
````

Another consequence, since an object can have several references:

````python
l2 = l
print(f"Identifier of the list l  : {id(l)}")
print(f"Identifier of the list l2 : {id(l)}")
````

and since lists are mutable, modifying the list ```l``` also modifies the list ```l2```:

````python
print(f"List l2  : {l2}")

l.append(2)

print(f"\nList l   : {l}")
print(f"List l2  : {l2}")
````
