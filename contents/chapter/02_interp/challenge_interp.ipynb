```python
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import plotly.io as pio
pio.templates.default = "seaborn"
import warnings
warnings.filterwarnings('ignore')
```

```python
def compute_divided_diff_coef(x, y):
    n = x.size
    coef = y.copy()
    for i in range(1,n):
        coef[i:] = (coef[i:] - coef[i-1:-1])/(x[i:] - x[:-i])
    return coef

def newton_interp(xi, yi, x):
    coef = compute_divided_diff_coef(xi, yi)
    n = xi.size
    p = np.zeros(x.size)
    for i in range(n-1,0,-1):
        p = (coef[i]+p) * (x-xi[i-1])
    p = p +  coef[0]
    return p
```

In this notebook, we show the result of the approximation of the function $f(x)=\sin(\pi x)$ on $[-1,1]$ with Newton's method, using 68 equidistributed interpolation points in $[-1,1]$.

```python
def f(x):
    return np.sin(np.pi*x)

xmin = -1.; xmax = 1.
```

## Polynomial approximation

```python
# x use to display f and pn
xplot = np.linspace(xmin, xmax, 500)

# maximal degree of Newton interpolating polynomial
n = 69

xk = np.linspace(xmin, xmax, n+1)
yk = f(xk)

pn = newton_interp(xk, yk, xplot)
    
# Create figure
fig = go.Figure(layout_title="Polynomial approximation with Newton's method")

# add f(x) plot
fig.add_trace(go.Scatter(visible=True, x=xplot, y=f(xplot), name="f(x)"))

# add yi and pn(x) invisible plots
fig.add_trace(go.Scatter(x=xplot, y=pn, name=f"p{n}(x)"))
fig.add_trace(go.Scatter(x=xk, y=yk, mode='markers', name="interpolation pts"))

# Create and add slider
fig.update_yaxes(range=[-3.1, 3.1])
legend = dict(orientation="h", y=1.1)
fig.update_layout(legend=legend)
fig.show()
```
