```python
import plotly.graph_objects as go
import numpy as np
import plotly.io as pio
pio.templates.default = "seaborn"
import warnings
warnings.filterwarnings('ignore')
```

```python
def plot(x, y1, y2):
    
    fig = go.Figure()

    fig.add_trace(go.Scatter(x=x, y=y1, name="y=x", legendgroup="g0",  legendgrouptitle_text=""))

    fig.add_trace(go.Scatter(x=x, y=y2, name="y=(1+alpha)x - alpha", marker_color="rgb(221,132,82)", legendgroup="g1",  legendgrouptitle_text="Original problem"))
    fig.add_trace(go.Scatter(x=[1], y=[1], name="exact solution", mode='markers', legendgroup="g1", marker=dict(symbol='x', size=9, color="rgb(221,132,82)")))

    epsilon = np.array([0.001, 0.01, 0.1])
    for i, eps_i in enumerate(epsilon):
        y_i = (1+alpha)*x - alpha - eps_i
        fig.add_trace(go.Scatter(x=x, y=y_i, name="y = (1 + alpha)x - alpha - eps", marker_color = 'rgb(85,168,104)',
                                 legendgroup="g2",  legendgrouptitle_text="Perturbed problem", visible=False))
        fig.add_trace(go.Scatter(x=[1+eps_i/alpha], y=[1+eps_i/alpha], name="perturbed solution", mode='markers',
                                 legendgroup="g2", marker=dict(symbol='x', size=9, color="rgb(85,168,104)"), visible=False))


    fig.data[3].visible = True
    fig.data[4].visible = True

    # Create and add slider
    steps = []
    for i, eps_i in enumerate(epsilon):
        args = [{"visible": [(el==2*i+3 or el==2*i+4 or el==0 or el==1 or el==2) for el in range(len(fig.data))]}]
        step = dict(method="update", label = f"{eps_i}",  args=args)
        steps.append(step)
    
    sliders = [dict(currentvalue={'prefix': 'epsilon = '}, steps=steps)]
    
    legend = dict(orientation="h", y=1.25)
    fig.update_layout(height=500, width=750, legend=legend, yaxis_range=[min(y1), max(y1)], sliders=sliders)
    fig.show()
```

## Linear system

We want to solve the following linear system:

$$
A=\left[
\begin{array}{lll}
-\gamma & 1  \\
- 1              & 1  \\
\end{array}
\right]
\begin{pmatrix}
     x    \\
     y            \\
\end{pmatrix}
= \begin{pmatrix}
     -\beta    \\
     0              \\
\end{pmatrix}
$$

whose solution is $x=\beta/(\gamma-1)$ and $y=\beta/(\gamma-1)$, corresponding to the coordinates of the intersection point of the lines $y=x$ and $y=\gamma\,x - \beta$. We will consider a case where $\gamma = 1+\alpha$ and $\beta = \alpha$, with $\alpha$ small compared to 1.

### Case where $\alpha$ is small

In the case where $\alpha$ is small compared to 1, as shown in the following figure where it equals 0.05, a small perturbation, for example of the right-hand side where $-\alpha$ becomes $-\alpha-\epsilon$, induces a modification of the solution that is large compared to the size of the perturbation, since the new solution still satisfies $\bar x=\bar y$ but $\bar x=1+\epsilon/\alpha$.

```python
alpha = 0.05
x = np.linspace(0.5, 3.5, 100)
y1 = x
y2 = (1+alpha)*x - alpha
plot(x, y1, y2)
```

In this setting, we can see that a perturbation of the right-hand side of order $\epsilon$ induces a modification of the solution of $\epsilon/\alpha$, which provides a lower bound in $1/\alpha$ on the conditioning that we will compute in the tutorial class and which is indeed of order $1/\alpha$. For $\alpha$ small compared to one, the conditioning is poor and we will be able to make the connection with Chapter 4 of the course, where the conditioning of a matrix will be presented.

### Case where $\alpha = -2$ 

```python
alpha = -2
x = np.linspace(0.5, 1.5, 100)
y1 = x
y2 = (1+alpha)*x - alpha
plot(x, y1, y2)
```

In this case, the conditioning of the problem is of order $1$ and a modification of the right-hand side induces a perturbation of order $\epsilon/2$. The problem is very well conditioned and this is solely associated with the mathematical structure of the original problem (matrix more or less close to a singular matrix) and in no way with an algorithmic evaluation issue.
