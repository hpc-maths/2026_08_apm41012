The Python module **mpmath** allows manipulating arbitrary-precision floating-point numbers by setting the size of the mantissa. The documentation of this module is available [here](http://mpmath.org/doc/current/) and the **mpmath** page [here](http://mpmath.org/).

```python
from mpmath import mp
```

## Initialization

To initialize an arbitrary-type floating-point number, simply use the `mpf` function: 

```python
a = mp.mpf(1.5)
print(a)
```

When initializing a floating-point number, it is possible and even preferable to provide a string to obtain better precision. For example, for the real number 0.1 which is not exactly representable in binary, it is preferable to use the string '0.1' rather than directly the Python constant 0.1 (which by default corresponds to a 64-bit floating-point number). It is also possible to use the string '1/10'. 

```python
b = mp.mpf('0.1')
print(b)
```

If we display the value of the variable `b` using the module's `nprint` function, we can specify the number of digits to display and verify that the value 0.1 is not exactly representable in binary :

```python
mp.nprint(b, n=19)
```

Which is not the case for *0.25*:  

```python
c = mp.mpf(0.25)
mp.nprint(c, n=19, strip_zeros=False)
```

because 0.25 is a fraction whose denominator is a power of 2 $(0.25 = 2^{-2})$. 

## Modifying the precision

By default, the size of the mantissa corresponds to that of double-precision floating-point numbers of the IEEE-754 standard, with a mantissa size of 53 bits:

```python
print(mp)
```

which corresponds to 15 significant digits (`mp.dps`) after conversion to decimal:

```python
print(mp.dps)
```

To increase the precision of floating-point numbers, simply set the size of the mantissa:

```python
mp.prec = 53
d = mp.mpf('1/10')
print()
print(f"For a floating-point number whose mantissa is {mp.prec} bits in size")
print(f"corresponding to {mp.dps} significant digits after conversion to decimal")
print(f"Display of the decimal representation using 40 digits")  
print(f"-> 1/10 = {mp.nstr(d, 40)} \n")      
      
mp.prec = 113
e = mp.mpf('1/10')
print(f"For a floating-point number whose mantissa is {mp.prec} bits in size")
print(f"corresponding to {mp.dps} significant digits after conversion to decimal")
print(f"Display of the decimal representation using 40 digits")  
print(f"-> 1/10 = {mp.nstr(e, 40)} \n")      
```

It is possible to clone an `mp` context so as to work with two precisions at the same time:

```python
mp.prec = 53
mp_def = mp.clone()
mp.prec = 113

f_dp = mp_def.mpf('1/10')
f_qp = mp.mpf('1/10')

g_dp = mp_def.mpf('1/8')
g_qp = mp.mpf('1/8')

h_dp = mp_def.mpf('1/3')
h_qp = mp.mpf('1/3')

print(f"Floating-point number with a  53-bit mantissa: 1/10 = {mp_def.nstr(f_dp, 40)}")
print(f"Floating-point number with a 113-bit mantissa: 1/10 = {mp.nstr(f_qp, 40)}\n")

print(f"Floating-point number with a  53-bit mantissa: 1/8 = {mp_def.nstr(g_dp, 40)}")
print(f"Floating-point number with a 113-bit mantissa: 1/8 = {mp.nstr(g_qp, 40)}\n")

print(f"Floating-point number with a  53-bit mantissa: 1/3 = {mp_def.nstr(h_dp, 40)}")
print(f"Floating-point number with a 113-bit mantissa: 1/3 = {mp.nstr(h_qp, 40)}")
```
