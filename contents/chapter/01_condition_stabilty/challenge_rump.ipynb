## Some surprising results...

We consider the function:

$$ f(x,y) = (333.75 - x^2) y^6 + x^2 (11 x^2 y^2 - 121y^4 - 2) + 5.5 y^8 + \frac{x}{2y} $$

```python
def f_rump(x,y):
    return (333.75-x*x)*y**6 + x**2*(11*(x**2)*(y**2) - 121*y**4 - 2) + 5.5*y**8 + x/(2*y)
```

Evaluating the function $f$ at the point $(x,y) = (77617, 33096)$ for different floating-point number formats gives the following results:

```python
# 32-bit floats 
mp.prec = 24
x = mp.mpf('77617')
y = mp.mpf('33096')
print("Evaluation of the function for 32-bit floats")
print(f"f(x,y) = {f_rump(x,y)}")

# 64-bit floats 
mp.prec = 53
x = mp.mpf('77617')
y = mp.mpf('33096')
print("\nEvaluation of the function for 64-bit floats")
print(f"f(x,y) = {f_rump(x,y)}")

# 128-bit floats 
mp.prec = 113
x = mp.mpf('77617')
y = mp.mpf('33096')
print("\nEvaluation of the function for 128-bit floats")
print(f"f(x,y) = {f_rump(x,y)}")
```

These results seem consistent : for each floating-point format we find the same result with a precision that increases with the size of the mantissa.

However, if we consider the following forms of the same function:

* $ f(x,y) = 333.75 y^6 + x^2 (11 x^2 y^2 - y^6 - 121y^4 - 2) + 5.5 y^8 + \displaystyle \frac{x}{2y} $

* $ f(x,y) = 333.75 yyyyyy + xx (11 xx \; yy - yyyyyy - 121yyyy - 2) + 5.5 yyyyyyyy + \displaystyle \frac{x}{2y} $

```python
def f_rump_02(x,y):
    return 333.75*y**6 + (x**2)*(11*(x**2)*(y**2) - y**6 - 121*y**4 - 2) + 5.5*y**8 + x/(2*y)

def f_rump_03(x,y):
    return 333.75*y*y*y*y*y*y + (x*x)*(11*(x*x)*(y*y) - y*y*y*y*y*y - 121*y*y*y*y - 2) + 5.5*y*y*y*y*y*y*y*y + x/(2*y)
```

Evaluating $f_2$ and $f_3$ at the point $(x,y) = (77617, 33096)$ gives the following results:

```python
# 32-bit floats 
mp.prec = 24
x = mp.mpf('77617')
y = mp.mpf('33096')
print("Evaluation of the function for 32-bit floats")
print(f"f2(x,y) = {f_rump_02(x,y)}")
print(f"f3(x,y) = {f_rump_03(x,y)}")

# 64-bit floats 
mp.prec = 53
x = mp.mpf('77617')
y = mp.mpf('33096')
print("\nEvaluation of the function for 64-bit floats")
print(f"f2(x,y) = {f_rump_02(x,y)}")
print(f"f3(x,y) = {f_rump_03(x,y)}")

# 128-bit floats 
mp.prec = 113
x = mp.mpf('77617')
y = mp.mpf('33096')
print("\nEvaluation of the function for 128-bit floats")
print(f"f2(x,y) = {f_rump_02(x,y)}")
print(f"f3(x,y) = {f_rump_03(x,y)}")
```

In fact, the exact value of the evaluation of the function $f$ at the point $(x,y) = (77617, 33096)$ is a rational number that can be obtained with the Python module **sympy**  and which equals -54767/66192 $\simeq$ -0.827396059946821.

```python
import sympy as sy
from sympy.abc import x, y

def f(x,y):
    return (sy.Rational(1335, 4)-x*x)*y**6 + x**2*(11*(x**2)*(y**2) - 121*y**4 - 2) + sy.Rational(11, 2)*y**8 + x*sy.Rational(1, 2*y) 

x = 77617
y = 33096

print("\nEvaluation of the function with sympy (using Rational method)")
print(f"f(x,y) = {f(x,y)}")
```

These results call for several remarks: 

* The choice of the method used to compute the function has a significant impact on the result, which can be radically different or similar for a quadruple-precision computation but wrong!
* The question then arises of understanding the origin of the errors obtained with single, double and quadruple precision.

**Challenge:** Propose an explanation that accounts for the results obtained.

To obtain the correct result, one has to use floating-point numbers whose mantissa size is at least equal to 122:

```python
# floats whose mantissa is equal to 122
mp.prec = 122
x = mp.mpf('77617')
y = mp.mpf('33096')
print(f"\nEvaluation of the function for floats whose mantissa is equal to {mp.prec}")
print(f"f(x,y)  = {f_rump(x,y)}")
print(f"f2(x,y) = {f_rump_02(x,y)}")
print(f"f3(x,y) = {f_rump_03(x,y)}")

print(f"\nNumber of significant digits with a mantissa of 122 : {mp.dps}")
```

**Elements of explanation**

The terms $a = 333.75 y^6 + x^2 (11 x^2 y^2 - y^6 - 121y^4 - 2)$ and $b = 5.5 y^8$ require a large precision in order to be represented exactly:

```python
mp.prec = 121
x = mp.mpf('77617')
y = mp.mpf('33096')
a = (333.75-x*x)*y**6 + x**2*(11*(x**2)*(y**2) - 121*y**4 - 2) 
b = 5.5*y**8
print(f"With a mantissa of {mp.prec} :")
print()
print(f"a = {a}")
print(f"a = {mp.nstr(a, 37)}")
print(f"b = {b}")
print(f"b = {mp.nstr(b, 37)}")
print(f"a+b = {a+b}")
print()

mp.prec = 122
x = mp.mpf('77617')
y = mp.mpf('33096')
a = (333.75-x*x)*y**6 + x**2*(11*(x**2)*(y**2) - 121*y**4 - 2) 
b = 5.5*y**8
print(f"With a mantissa of {mp.prec} :")
print()
print(f"a = {a}")
print(f"a = {mp.nstr(a, 37)}")
print(f"b = {b}")
print(f"b = {mp.nstr(b, 37)}")
print(f"a+b = {a+b}")
```

In fact, we are adding very large decimal numbers with 36 significant digits, and these digits absolutely have to be evaluated in order for the sum of $a$ and $b$ to be exactly $-2$. With one significant digit less, the result is $0$, as shown by the previous evaluation. 

The question that then arises is whether the problem comes from the function to be evaluated or from the method used to compute it. An exercise in the PC (tutorial class) will show that the evaluation of the function at this point is extremely ill-conditioned and that, in order to obtain a forward error that is not catastrophic compared with the result, a very small machine precision, and thus a large number of significant digits, is required.
