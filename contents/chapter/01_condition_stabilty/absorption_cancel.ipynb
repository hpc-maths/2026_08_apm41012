The goal of this challenge is to provide some explanations about the absorption and cancellation errors that are classical in the field and to properly understand the link with the machine representation and the round-off (machine) error. For the purposes of the challenge, we will work with 6 significant digits and use the mpmath library.

```python
mp.dps = 6
print(f"Machine epsilon for floats with {mp.dps} significant digits : {mp.eps}")
```

## Absorption error

Let $a = 1.23456$, $b_1 = 0.00000123456$ and $b_2 = 0.0000012$ and consider the sum of $a$ and $b_1$ then the sum of $a$ and $b_2$ for floats with 6 significant digits. For presentation purposes, we represent the decimal numbers with 12 digits associated with the machine representations at this precision. 

```python
mp.dps = 6 

a  = mp.mpf('1.23456')
b1 = mp.mpf('0.00000123456')
b2 = mp.mpf('0.0000012')

res1 = a+b1
res2 = a+b2

print("Using floats with 6 significant digits :")
print("the decimal number associated with the machine representation of a + b1 : ", mp.nstr(res1, n=12))
print("the decimal number associated with the machine representation of a + b2 : ", mp.nstr(res2, n=12))

ref = 1.23456 + 0.00000123456
err1 = np.abs(float(res1) - ref)/ref
err2 = np.abs(float(res2) - ref)/ref
print()
print("Relative error associated with the computation of a + b1 : ", err1)
print("Relative error associated with the computation of a + b2 : ", err2)
```

## Cancellation error

Let $a = 1.23456$ and $b = 1.23454$ and consider the difference of $a$ and $b$ for floats with 6 significant digits.

```python
mp.dps = 6
print("Using floats with 6 significant digits :")

a = mp.mpf(1.23456)
print("the decimal number associated with the machine representation of a   is ", mp.nstr(a, n=12))

b = mp.mpf(1.23454)
print("the decimal number associated with the machine representation of b   is ", mp.nstr(b, n=12))

res = a-b
print("the decimal number associated with the machine representation of a-b is ", mp.nstr(res, n=12))

ref = 1.23456 - 1.23454

err = abs(float(res) - ref)/ref
print("\nRelative error :", err)
```
