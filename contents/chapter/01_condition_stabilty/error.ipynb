```python
from mpmath import mp
```

One consequence of using floating-point arithmetic is the loss of certain mathematical properties. For example, addition is no longer associative. 

We consider the sum:

$$ s_1 = 1  + \frac{1}{2} + \frac{1}{3} + \dots + \frac{1}{10^6} $$

```python
# we set the precision of the floats
mp.prec = 24

s1 = mp.mpf('0')
for i in range(1,1_000_001):
    s1 += 1/i
print(f"s1 with {mp.dps} significant digits = {s1}")
```

then the same sum evaluated in reverse order:

$$ s_2 = \frac{1}{10^6} + \dots  + \frac{1}{3} + \frac{1}{2}  + 1 $$

```python
s2 = mp.mpf('0')
for i in range(1_000_000, 0, -1):
    s2 += 1/i
print(f"s2 with {mp.dps} significant digits = {s2}")

print(f"\nRelative difference between the two sums with {mp.dps} significant digits: {abs(s1-s2)/s2 * 100} %")
```

We can compute a reference value using high precision. We check that at this precision the computation of the sum does not depend on the order of the terms:

```python
mpref = mp.clone()
mpref.prec = 113
sref1 = mpref.mpf('0')
for i in range(1_000_000, 0, -1):
    sref1 += 1/i


sref2 = mpref.mpf('0')
for i in range(1,1_000_001):
    sref2 += 1/i

print(f"sref1 (decreasing order) with {mpref.dps} significant digits = {sref1} \n")
print(f"sref2 (increasing order)   with {mpref.dps} significant digits = {sref2} \n")

print(f"sref2 - sref1 = {sref1-sref2}")
```

The reader may also compute this exact sum in the field of rational numbers with Sagemath and show that converting the resulting number into a double-precision real is consistent with the value obtained in quadruple precision.

We can then evaluate the error made between the computation of the 2 sums at the initial precision:

```python
print(f"1st form: error with {mp.dps} significant digits = {abs(s1-sref1)/sref1 * 100} %")
print(f"2nd form: error with {mp.dps} significant digits = {abs(s2-sref1)/sref1 * 100} %")
```

In conclusion, the second algorithm is much more backward stable and therefore leads to a much smaller relative forward error. There is thus a more "relevant" way of performing the operations when one knows how real numbers are represented on a machine, by starting with the addition of small numbers and finishing with the large ones when the sum is already significant.
