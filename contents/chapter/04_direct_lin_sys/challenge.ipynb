```python
import numpy as np
from numpy import linalg
```

```python
def pretty_print_vec(v):
    for vi in v: print(vi)
    return
```

We consider the following system:

$$
\begin{pmatrix}
 1/2 & 1/3 & 1/4 & 1/5\\
 1/3 & 1/4 & 1/5 & 1/6\\
 1/4 & 1/5 & 1/6 & 1/7\\
 1/5 & 1/6 & 1/7 & 1/8
\end{pmatrix}
\begin{pmatrix}
 x_1 \\
 x_2 \\ 
 x_3 \\ 
 x_4 \\ 
\end{pmatrix}
=
\begin{pmatrix}
 3511/13860 \\
 277/1540 \\ 
 40877/291060 \\
 3203/27720
\end{pmatrix}
$$

whose exact solution is:

$$
\begin{pmatrix}
 1/3 \\
 1/11 \\ 
 1/9 \\
 1/7
\end{pmatrix}
$$



We solve the system using 64-bit floating-point numbers:

```python
xexa64 = np.array([1/3, 1/11, 1/9, 1/7])
print("Exact solution:")
pretty_print_vec(xexa64)

# double precision
A64 = np.array([[1/2, 1/3, 1/4, 1/5], 
               [1/3, 1/4, 1/5, 1/6],
               [1/4, 1/5, 1/6, 1/7], 
               [1/5, 1/6, 1/7, 1/8]], dtype=np.float64)

b64 = np.array([3511/13860, 277/1540, 40877/291060, 3203/27720], dtype=np.float64)

x64 = np.linalg.solve(A64, b64)

print("\nSolution with 64-bit floating-point numbers:")
pretty_print_vec(x64)

####pretty_print_vec(np.abs(xexa64 - x64))

res64 = np.abs(np.dot(A64,x64)-b64)
print("\nResidual |Ax-b|:")
pretty_print_vec(res64)
```

Then, we solve the system using 32-bit floating-point numbers:

```python
xexa32 = np.array([1/3, 1/11, 1/9, 1/7], dtype=np.float32)

print("Exact solution:")
pretty_print_vec(xexa32)

# simple precision
A32 = np.array([[1/2, 1/3, 1/4, 1/5], 
                 [1/3, 1/4, 1/5, 1/6],
                 [1/4, 1/5, 1/6, 1/7], 
                 [1/5, 1/6, 1/7, 1/8]], dtype=np.float32)

b32 = np.array([3511/13860, 277/1540, 40877/291060, 3203/27720], dtype=np.float32)

x32 = np.linalg.solve(A32, b32)

print("\nSolution with 32-bit floating-point numbers:")
pretty_print_vec(x32)

#pretty_print_vec(np.abs(xexa32 - x32))

res32 = np.dot(A32,x32)-b32
print("\nResidual Ax-b:")
pretty_print_vec(res32)
```

The results show that only 3 to 4 decimal places remain correct when going from double precision to single precision. Yet the residuals of the solutions $Ax - b$ are of the order of the chosen machine precision. What is the origin of the error obtained?
