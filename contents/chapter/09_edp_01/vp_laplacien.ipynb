```python
import plotly.graph_objects as go
import numpy as np
```

Based on the classical analysis of the spectrum of the matrix $A$, we recall that the eigenvalues are given by:

$$
\lambda_k = - \frac{4}{(\Delta x)^2}\,\sin^2\left( \frac{k\,\pi}{2(n+1)}\right),\quad k=0,\ldots,n-1.
$$

The classical analysis gives the following eigenvalues of the Laplacian on an interval of length $L$:

$$
\frac{-k^2 \pi^2}{L^2}, \quad k=0,\ldots,n-1,$$

```python
def plot_eigen_values(lx, nx):
    
    dx = lx /(nx+1)

    k = np.arange(nx)
    
    cont_eig_vals = -((k * np.pi) / lx)**2
    disc_eig_vals = -(4/(dx*dx)) * np.sin((k*np.pi)/(2*(nx+1)))**2   
    
    fig = go.Figure()

    fig.add_trace(go.Scatter(x=k, y=cont_eig_vals, name="Eigenvalues of the Laplacian"))
    fig.add_trace(go.Scatter(x=k, y=disc_eig_vals, name="Eigenvalues of the discretized Laplacian"))

    # Create and add slider
    steps = []
    for nxi in np.arange(1000,10001,100):
        dx = lx /(nxi+1)
        k = np.arange(nxi)
        cont_eig_vals = -((k * np.pi) / lx)**2
        disc_eig_vals = -(4/(dx*dx)) * np.sin((k*np.pi)/(2*(nxi+1)))**2   
        step = dict(method="update", label = f"{nxi}", args=[{"x": [k, k], "y": [cont_eig_vals, disc_eig_vals]}])
        steps.append(step)
    sliders = [dict(currentvalue={'prefix': 'nx = '}, steps=steps)]

    fig.update_layout(sliders=sliders, yaxis=dict(exponentformat='e'),  height=600,
                      legend=dict(x=0.65, y=0.99, bgcolor='rgba(0,0,0,0)'))
    fig.show()
```

```python
xmin = -5.
xmax = 5.
lx = xmax-xmin
nx = 1001

plot_eigen_values(lx, nx)
```
