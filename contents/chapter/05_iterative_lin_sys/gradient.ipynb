```python
import numpy as np
from matplotlib import pyplot as plt
```

We consider a $2\times 2$ system for which the behaviour of the algorithms can easily be visualised:

$$
A=\begin{pmatrix}
7 & 0 \\
0 & 1
\end{pmatrix},\qquad b=\begin{pmatrix}
2 \\ -1
\end{pmatrix}
$$

```python
a = np.array([[7., 0.], [0., 1.]])

b = np.array([2., -1.])

xexa = np.linalg.solve(a,b)
```

```python
def f(a, b, x1, x2):
    return 0.5*((a[0,0]*x1+a[0,1]*x2)*x1 + (a[1,0]*x1+a[1,1]*x2)*x2) - b[0]*x1 - b[1]*x2
```

## Gradient method with fixed step

```python
alpha = 0.28

# initialization
nit_max = 100
xk = [np.zeros(2)]
rk = np.dot(a,xk[0]) - b
# gradient iteration
for k in range(nit_max):
    xk.append(xk[k] - alpha*rk)
    rk = np.dot(a,xk[k+1]) - b
    if (np.linalg.norm(rk) < 1.e-6): break
        
xk = np.array(xk)
        
print("Gradient method with fixed step:")
print(f"-> number of iterations {k+1}")

x1 = np.linspace(-0.5, 1., 200)
x2 = np.linspace(-2.5, 0.5, 200)
x1, x2 = np.meshgrid(x1, x2)

z = f(a, b, x1, x2)

nlevel = 8
level =  f(a, b, xk[0:nlevel,0], xk[0:nlevel,1])

plt.figure(figsize=(7.5,15))
plt.contour(x1, x2, z, np.flip(level))
plt.plot(xk[0:nlevel,0], xk[0:nlevel,1], '--ro')
plt.scatter(xexa[0], xexa[1])
plt.show()
```

## Gradient method with optimal step

```python
# initialization
nit_max = 100
xk = [np.zeros(2)]
rk = np.dot(a,xk[0]) - b
# gradient iteration
for k in range(nit_max):
    alpha = np.dot(rk,rk)/ np.dot(np.dot(a,rk),rk)
    xk.append(xk[k] - alpha*rk)
    rk = np.dot(a,xk[k+1]) - b
    if (np.linalg.norm(rk) < 1.e-6): break

xk = np.array(xk)

print("Gradient method with optimal step:")
print(f"-> number of iterations {k+1}")

x1 = np.linspace(-0.5, 1., 200)
x2 = np.linspace(-2.5, 0.5, 200)
x1, x2 = np.meshgrid(x1, x2)

z = f(a, b, x1, x2)

nlevel = 8
level =  f(a, b, xk[0:nlevel,0], xk[0:nlevel,1])

plt.figure(figsize=(7.5,15))
plt.contour(x1, x2, z, np.flip(level))
plt.plot(xk[0:nlevel,0], xk[0:nlevel,1], '--ro')
plt.scatter(xexa[0], xexa[1])
plt.show()
```

## Conjugate gradient

```python
# initialization
xk = [np.zeros(2)]
rk = b - np.dot(a,xk[0])
pk = rk
rkm1 = rk
# gradient iteration
for k in range(b.size):
    apk = np.dot(a,pk)
    alpha = np.dot(rk,rk)/ np.dot(pk, np.dot(a,pk))
    xk.append(xk[k] + alpha*pk)
    rk = rk - alpha*apk
    if (np.linalg.norm(rk) < 1.e-6): break
    beta = np.dot(rk,rk) / np.dot(rkm1, rkm1)
    pk = rk + beta*pk
    rkm1 = rk

xk = np.array(xk)

print("Conjugate gradient:")
print(f"-> number of iterations {k+1}")

x1 = np.linspace(-0.5, 1., 200)
x2 = np.linspace(-2.5, 0.5, 200)
x1, x2 = np.meshgrid(x1, x2)

z = f(a, b, x1, x2)

nlevel = 3
level =  f(a, b, xk[0:nlevel,0], xk[0:nlevel,1])

plt.figure(figsize=(7.5,15))
plt.contour(x1, x2, z, np.flip(level))
plt.plot(xk[0:nlevel,0], xk[0:nlevel,1], '--ro')
plt.scatter(xexa[0], xexa[1])
plt.show()
```
